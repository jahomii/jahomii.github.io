The following program will take a CSV of spotify streaming history, and use it to determine your most popular artists and songs. 

NOTE: All "install.packages" lines will be commented out, if at any point something is not running due to an "object does not exist" error or similar, uncomment and install the required package. 

```r
# confirm that the file exists
file.exists("spotify.csv")
```

```r
# creating a data frame using the csv
streaming_history <- read.csv(file = "spotify.csv")

# viewing the first couple rows to ensure everything looks correct
str(streaming_history)
```

```r
# summary can tell us a bit about each variable.
summary(streaming_history)
```

Did you notice that ms_played has a min of 0? I am not interested in counting songs with 0 ms_played as individual listens. Lets remove those. 

I also have a max ms_played of 3727766 which roughly translates to.... 62 minutes? That's not right. I'll be setting a listening cap of 15 minutes to account for some of the extra long songs I listen to. Looking at you, Guns n' Roses. 

I am also going to remove any songs where I stopped listening after 20000 ms, since that means I only listened for about 20 seconds.

```r
# if any songs play for less than 2000 ms or over 900000 ms (15 mins) mark them NA
streaming_history$ms_played <- ifelse(streaming_history$ms_played > 20000 & streaming_history$ms_played < 900000, streaming_history$ms_played, NA)
# remove null values
cleaned_streaming_history <- na.omit(streaming_history)
summary(streaming_history)
summary(cleaned_streaming_history)
```

as we can see by these two summaries, the length of all my variables dropped from 157779 to 64523. The size of my dataset has decreased significantly. Normally this is a bad sign but I knew going into this that this dataset was going to be very inconsistent (since its based on the whims of my own listening habits) so for this instance, its okay. Lets move on. 

```r
#install.packages("data.table")
library(data.table)

# converting streaming history into a data table
dt_streaming = data.table(cleaned_streaming_history)
```

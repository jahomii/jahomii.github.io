## Looks like some cleaning is required
Did you notice that ms_played has a min of 0? I am not interested in counting songs with 0 ms_played as individual listens. Lets remove those. 

I also have a max ms_played of 3727766 which roughly translates to.... 62 minutes? That's not right. I'll be setting a listening cap of 15 minutes to account for some of the extra long songs I listen to. Looking at you, Guns n' Roses. 

I am also going to remove any songs where I stopped listening after 20000 ms, since that means I only listened for about 20 seconds.

```r
# if any songs play for less than 2000 ms or over 900000 ms (15 mins) mark them NA
streaming_history$ms_played <- ifelse(streaming_history$ms_played > 20000 & streaming_history$ms_played < 900000, streaming_history$ms_played, NA)
# remove null values
cleaned_streaming_history <- na.omit(streaming_history)
summary(streaming_history)
summary(cleaned_streaming_history)
```

## As we can see...
The length of all my variables dropped from 157779 in the first summary to 64523 in the second. The size of my dataset has decreased significantly. Normally this is a bad sign but I knew going into this that this dataset was going to be very inconsistent (since its based on the whims of my own listening habits) so for this instance, its okay. Lets move on.

## Lets discover how likely I am to skip certain songs. 

We'll be using the OneR package for machine learning, and we'll specifically use the streaming_history dataset. This dataset does not have songs under 20 seconds long removed, since it will otherwise assume I dont skip very many songs at all.

```r
# mild cleaning, I dont want songs to be included that I haven't listened to at LEAST 10 times
skipped_songs <- streaming_history[streaming_history$song_title %in% names(which(table(streaming_history$song_title) >= 10)), ]

# ensure all variables are factors
skipped_songs[] <- lapply(skipped_songs, function(x) {
  if (is.character(x) | is.logical(x)) factor(x) else x
})

# removing columns that aren't neccesary
skipped_songs$ts = NULL
skipped_songs$ms_played = NULL
skipped_songs$track_url = NULL
skipped_songs$reason_start = NULL
skipped_songs$reason_end = NULL
skipped_songs$offline = NULL
skipped_songs$shuffle = NULL
skipped_songs$album_name = NULL

head(skipped_songs)
```

```r
# install.packages("OneR")
library(OneR)

# using the OneR model to get my listening rules
skipped_songs_1R <- OneR(skipped ~ song_title, data = skipped_songs)
skipped_songs_1R
```

```r
skipped_songs_pred <- predict(skipped_songs_1R, skipped_songs)
table(actual = skipped_songs$skipped, predicted = skipped_songs_pred)
```

For songs, we have 77.83% accuracy

- [True Positive] correctly predicted skipped = 14,022
- [True Negative] correctly predicted NOT skipped = 99,185
- [False Positive] incorrectly predicted skipped = 7,385
- [False Negative] incorrectly predicted NOT skipped = 24,657

let's do the same process with artists

```r
# mild cleaning, I dont want artists to be included that I haven't listened to at LEAST 3 times
skipped_artists <- streaming_history[streaming_history$artist_name %in% names(which(table(streaming_history$artist_name)>=3)), ]

skipped_artists[] <- lapply(skipped_artists, function(x) {
  if (is.character(x) | is.logical(x)) factor(x) else x
})

# removing collumns that aren't neccesary
skipped_artists$ts = NULL
skipped_artists$ms_played = NULL
skipped_artists$track_url = NULL
skipped_artists$reason_start = NULL
skipped_artists$reason_end = NULL
skipped_artists$offline = NULL
skipped_artists$shuffle = NULL
skipped_artists$album_name = NULL

head(skipped_artists)
```

```r
skipped_artists_1R <- OneR(skipped ~ artist_name, data = skipped_artists)
skipped_artists_1R
```

```r
skipped_artists_pred <- predict(skipped_artists_1R, skipped_artists)
table(actual = skipped_artists$skipped, predicted = skipped_artists_pred)
```

For artists, we have 76.58% accuracy

- [True Positive] correctly predicted skipped = 11,209
- [True Negative] correctly predicted NOT skipped = 107,182
- [False Positive] incorrectly predicted skipped = 6,431
- [False Negative] incorrectly predicted NOT skipped = 29,771
